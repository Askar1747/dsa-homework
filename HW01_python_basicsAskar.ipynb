1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.

````python
# База
cities_clean=[]
amounts_int= []
for a in cities:
    s= a.strip()
    v=s.lower()
    aq=v.title()
    cities_clean.append(aq)
cities_clean
````

````python
for b in amounts:
    amounts_int.append(int(b))
````

````python
for c in amounts:
    amounts_int.append(int(c))
len_amounts= len(amounts)
mean_value= round(sum(amounts_int)/len(amounts_int))
sum_amounts= sum(amounts_int)
````

````python
for city, amount, apaid in zip(cities_clean, amounts_int, paid):
    if apaid == False:
        print(f"{city} , {amount} , {apaid}")
````

````python
CITY = []

for city in cities_clean:
    if city not in CITY:
        CITY.append(city)

for city in CITY:
    total = 0

    for c, amount, apaid in zip(cities_clean, amounts_int, paid):
        if c == city and apaid == True:
            total += amount

    print(f"{city} | {total}")
````

````python
revenue_ast= 0
revenue_alt=0
for city, amount, apaid in zip(cities_clean, amounts_int, paid):
    if city == "Almaty" and apaid == True:
        revenue_alt += amount
    elif city == "Astana" and apaid == True:
        revenue_ast += amount
print(f"Выручка по Алмату: {revenue_alt}")
print(f"Выручка по Астана: {revenue_ast}")
````

````python
max_amount =0
max_index = 0
for index, amount in enumerate(amounts_int):
    if amount > max_amount:
        max_amount = amount
        max_index = index
print(f"Номер заказа с самым большим чеком: {max_index}")
````
