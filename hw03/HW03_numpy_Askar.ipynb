````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

# Стандарт

Маркетинг предлагает скидку 15% на оплаченные заказы дороже 50 000 ₸. 
Сколько выручки магазин недополучит? Посчитайте дважды: со всеми заказами и без заказов клиента TEST. Почему ответы различаются в несколько раз?

````python
mask_all = paid & (amount > 50000)
loss_all = (amount[mask_all] * 0.15).sum()
mask_clean_test = paid & (amount > 50000) & (cust != "TEST")
loss_without_test = (amount[mask_clean_test] * 0.15).sum()
````

````python
print(round(loss_all,2))
print(round(loss_without_test,2))
print(round(loss_all-loss_without_test,2))
````

Для каждого города — доля заказов, доставленных за 1–2 дня. Список городов: np.unique(city); цикл по 6 городам здесь можно.

````python
delivered = ~np.isnan(days)
cities = np.unique(city)
for c in cities:
    total = np.sum(city == c)
    delivered_count = np.sum((city == c) & delivered)

    share = delivered_count / total * 100

    print(c, ":", round(share, 2), "%")
````

Сравните скорость: сумма оплаченных заказов циклом for и через маску. Во сколько раз быстрее?

````python
import time
start = time.perf_counter()
total_loop = 0
for i in range(len(amount)):
    if paid[i]:
        total_loop += amount[i]
time_loop = time.perf_counter() - start
start = time.perf_counter()
total_mask = amount[paid].sum()
time_mask = time.perf_counter() - start
````

````python
print(total_loop)
print(total_mask)
print(time_loop)
print(time_mask)
print(round(time_loop / time_mask, 2), )
````
