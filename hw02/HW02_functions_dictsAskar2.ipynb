````python
def to_num(x):
    try:
        return float(x)
    except:
        return 0


def clean_city(x):
    return x.strip().title()
````

````python
for row in rows:
    row["amount"] = to_num(row["amount"])
    row["city"] = clean_city(row["city"])
````

````python
categories = []

for row in rows:
    category = row["category"]

    if category not in categories:
        categories.append(category)


category_counts = []

for category in categories:
    count = 0

    for row in rows:
        if row["category"] == category:
            count += 1

    category_counts.append([category, count])


# Сортировка по убыванию
for i in range(len(category_counts)):
    for j in range(i + 1, len(category_counts)):
        if category_counts[i][1] < category_counts[j][1]:
            category_counts[i], category_counts[j] = category_counts[j], category_counts[i]


print("Количество заказов по категориям:")

for item in category_counts:
    print(item[0], "-", item[1])
````

````python
categories = []

for row in rows:
    if row["category"] not in categories:
        categories.append(row["category"])


for category in categories:
    total = 0
    count = 0

    for row in rows:
        if row["category"] == category and row["status"] == "paid":
            total += row["amount"]
            count += 1

    if count > 0:
        average = total / count

        print("Категория:", category)
        print("Выручка:", total)
        print("Средний чек:", average)
        print("-" * 30)
````

## [Стандарт]
5. Выручка оплаченных по месяцам и изменение октября к сентябрю в процентах.
6. Доля отменённых заказов по каждому (чистому) городу, в процентах с одним знаком.
7. Топ-5 клиентов (`customer_id`) по выручке оплаченных заказов. Кто на первом месте и что с ним не так? Ответьте в Markdown-ячейке.
8. Сколько **уникальных** клиентов в каждом городе? Подсказка: словарь множеств.

````python
monthly_revenue = {}

for row in rows:
    if row["status"] == "оплачен":
        month = row["order_date"][:7]

        if month in monthly_revenue:
            monthly_revenue[month] += row["amount"]
        else:
            monthly_revenue[month] = row["amount"]


# Вывод выручки по месяцам
for month in monthly_revenue:
    print(month, "-", monthly_revenue[month])


# Сравнение сентября и октября
september = monthly_revenue.get("2026-09", 0)
october = monthly_revenue.get("2026-10", 0)

if september != 0:
    change = (october - september) / september * 100

    print("Изменение октября к сентябрю:",
          round(change, 1), "%")
else:
    print("Сентябрь отсутствует или выручка равна 0")
````

````python
city_total = {}
city_cancelled = {}

for row in rows:
    city = row["city"]

    # Общее количество заказов в городе
    if city in city_total:
        city_total[city] += 1
    else:
        city_total[city] = 1

    # Количество отменённых заказов
    if row["status"] == "отменён":
        if city in city_cancelled:
            city_cancelled[city] += 1
        else:
            city_cancelled[city] = 1


# Расчёт процента отмен
for city in city_total:
    if city in city_cancelled:
        cancelled = city_cancelled[city]
    else:
        cancelled = 0

    percent = cancelled / city_total[city] * 100

    print(city, "-", round(percent, 1), "%")
````

````python
customer_revenue = {}

for row in rows:
    if row["status"] == "оплачен":
        customer = row["customer_id"]

        if customer in customer_revenue:
            customer_revenue[customer] += row["amount"]
        else:
            customer_revenue[customer] = row["amount"]
top_customers = list(customer_revenue.items())
for i in range(len(top_customers)):
    for j in range(i + 1, len(top_customers)):
        if top_customers[i][1] < top_customers[j][1]:
            top_customers[i], top_customers[j] = top_customers[j], top_customers[i]
# Вывод топ-5
for i in range(5):
    print(top_customers[i][0], "-", top_customers[i][1])
````

Клиент с самым большим объёмом оплаченных покупок занимает первое место по выручке. Его сумма покупок значительно превышает показатели остальных клиентов. Необходимо проверить, нет ли у него аномально крупных заказов или ошибок в данных.

````python
cities = []

for row in rows:
    if row["city"] not in cities:
        cities.append(row["city"])

print(cities)
````
